# 🚀 Proxy-Core Windows编译器

这个笔记本用于在Google Colab上交叉编译proxy-core为Windows可执行文件。

## 特性

- ✅ ECH (Encrypted Client Hello) 原生支持
- ✅ Yamux多路复用
- ✅ SOCKS5和HTTP代理
- ✅ TLS指纹随机化
- ✅ CDN优选IP支持

## 使用步骤

1. 运行所有单元格
2. 等待编译完成（约10-15分钟）
3. 下载生成的.exe文件
4. 在Windows上运行

## 📦 步骤1: 克隆仓库

````python
%%bash
# 克隆jarustls仓库
echo "📥 Cloning repository..."
git clone https://github.com/briomianopc/jarustls.git
cd jarustls
echo "✅ Repository cloned"
echo ""
echo "📂 Repository structure:"
ls -la
````

## 🔧 步骤2: 安装Rust和工具链

````python
%%bash
# 安装Rust
echo "🦀 Installing Rust..."
curl --proto '=https' --tlsv1.2 -sSf https://sh.rustup.rs | sh -s -- -y
source $HOME/.cargo/env

# 验证安装
rustc --version
cargo --version

echo "✅ Rust installed successfully"
````

## 🎯 步骤3: 添加Windows目标

````python
%%bash
source $HOME/.cargo/env

echo "🎯 Adding Windows target..."
rustup target add x86_64-pc-windows-gnu

echo "✅ Windows target added"
echo ""
echo "📋 Installed targets:"
rustup target list --installed
````

## 🔨 步骤4: 安装MinGW交叉编译器

````python
%%bash
echo "🔨 Installing MinGW..."
apt-get update -qq
apt-get install -y -qq mingw-w64

echo "✅ MinGW installed"
echo ""
echo "🔍 Verifying installation:"
x86_64-w64-mingw32-gcc --version | head -1
````

## ⚙️ 步骤5: 配置Cargo

````python
%%bash
source $HOME/.cargo/env

echo "⚙️  Configuring Cargo..."
mkdir -p ~/.cargo

cat > ~/.cargo/config.toml << 'EOF'
[target.x86_64-pc-windows-gnu]
linker = "x86_64-w64-mingw32-gcc"
ar = "x86_64-w64-mingw32-ar"
EOF

echo "✅ Cargo configured"
echo ""
echo "📄 Config file:"
cat ~/.cargo/config.toml
````

## 🏗️ 步骤6: 编译proxy-core

⏱️ 这一步需要10-15分钟，请耐心等待...

````python
%%bash
source $HOME/.cargo/env

cd jarustls/proxy-core

echo "🏗️  Building proxy-core for Windows..."
echo "This will take 10-15 minutes..."
echo ""

# 编译
cargo build --release --target x86_64-pc-windows-gnu

# 检查结果
if [ -f "target/x86_64-pc-windows-gnu/release/proxy-core.exe" ]; then
    echo ""
    echo "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
    echo "✅ Build successful!"
    echo "━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━"
    echo ""
    echo "📁 Output file:"
    ls -lh target/x86_64-pc-windows-gnu/release/proxy-core.exe
    echo ""
    echo "📊 File size:"
    du -h target/x86_64-pc-windows-gnu/release/proxy-core.exe
    echo ""
    echo "🔍 File type:"
    file target/x86_64-pc-windows-gnu/release/proxy-core.exe
else
    echo "❌ Build failed"
    exit 1
fi
````

## 📦 步骤7: 准备下载文件

````python
%%bash
cd jarustls/proxy-core

echo "📦 Preparing download..."

# 复制到根目录方便下载
cp target/x86_64-pc-windows-gnu/release/proxy-core.exe /content/proxy-core-windows.exe

echo "✅ File ready for download"
echo ""
echo "📁 Download location: /content/proxy-core-windows.exe"
ls -lh /content/proxy-core-windows.exe
````

## 📥 步骤8: 下载文件

运行下面的代码单元格，然后点击下载链接。

````python
from google.colab import files

print("📥 Downloading proxy-core-windows.exe...")
print("")
print("⏬ Click the download link below:")
files.download('/content/proxy-core-windows.exe')
print("")
print("✅ Download started!")
````

## 📋 使用说明

### 在Windows上使用

1. **下载完成后**，将`proxy-core-windows.exe`重命名为`proxy-core.exe`

2. **创建配置文件** `config.toml`:
```toml
[server]
address = "your-server.com"
port = 443
uuid = "your-uuid-here"
ws_path = "/"

[tls]
randomize_fingerprint = true
enable_ech = true
sni = "your-server.com"

[proxy]
socks5_enabled = true
socks5_bind = "127.0.0.1"
socks5_port = 1080

http_enabled = true
http_bind = "127.0.0.1"
http_port = 8080
```

3. **运行**:
```cmd
proxy-core.exe --server-domain your-server.com --target 1.1.1.1 --uuid your-uuid
```

4. **测试**:
```cmd
curl --socks5 127.0.0.1:1080 https://www.google.com
```

### 命令行参数

```
--server-domain <DOMAIN>    服务器域名（必需）
--target <IP/CNAME>         优选IP或CNAME（可选）
--port <PORT>               服务器端口（默认443）
--uuid <UUID>               认证UUID
--enable-ech <BOOL>         启用ECH（默认true）
--enable-fingerprint-randomization <BOOL>  启用指纹随机化（默认true）
--socks5-bind <ADDR>        SOCKS5监听地址（默认127.0.0.1:1080）
--http-bind <ADDR>          HTTP监听地址（默认127.0.0.1:8080）
--log-level <LEVEL>         日志级别（默认info）
```

### 查看帮助

```cmd
proxy-core.exe --help
```
